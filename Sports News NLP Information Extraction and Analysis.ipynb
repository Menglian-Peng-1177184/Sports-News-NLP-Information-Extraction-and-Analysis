# Loading articles and extracting noun phrases/chunking

```python
!pip install goose3
from goose3 import Goose
import spacy

g = Goose()
article1 = g.extract(url='https://www.theguardian.com/sport/2026/aug/08/australia-wallabies-japan-nations-championship-2026-match-report')
article2 = g.extract(url='https://www.theguardian.com/sport/2026/jul/18/argentina-england-nations-championship-rugby-union-match-report')
article3 = g.extract(url='https://www.theguardian.com/sport/2026/jul/04/south-africa-england-nations-championship-rugby-union-match-report')
article4 = g.extract(url='https://www.theguardian.com/sport/live/2026/jul/11/fiji-v-england-nations-championship-rugby-union-live')
article5 = g.extract(url='https://www.theguardian.com/football/2026/jul/19/spain-win-world-cup-ferran-torres-10-man-argentina-extra-time')
print(article1.title)
print(article2.title)
print(article3.title)
print(article4.title)
print(article5.title,"\n")

articles = [article1.cleaned_text, article2.cleaned_text,article3.cleaned_text,article4.cleaned_text,article5.cleaned_text]
whole_articles = " ".join(articles)
nlp = spacy.load('en_core_web_sm')
doc = nlp(whole_articles)
for chunk in doc.noun_chunks:
    print(chunk.text, chunk.label_, chunk.root.text)
```

```python
def printDocDep(doc):
  for token in doc:
    print("{0}/{1} <--{2}-- {3}/{4}".format(token.text, token.tag_, token.dep_, token.head.text, token.head.tag_))
printDocDep(doc)
```

# Display a dependency tree.

```python
from spacy import displacy
displacy.render(doc, style='dep', jupyter = True, options={'distance': 100})
```

#Extracting clauses using tagged text.

```python
from spacy.symbols import nsubj, VERB, dobj, pobj, NOUN,PROPN, PERSON

persons, nouns,verbs = [],[],[]
subjs, dobjs, headVerbs = [],[],[]

for token in doc:

    if token.pos == NOUN or token.pos == PROPN: nouns.append(token)
    if token.pos == VERB: verbs.append(token)
    if token.ent_type == PERSON: persons.append(token)

    if token.dep == nsubj: subjs.append(token)
    if token.dep == dobj: dobjs.append(token)

    if token.dep == nsubj and token.head.pos == VERB:
        headVerbs.append(token.head)

    if token.dep == nsubj and token.head.pos == VERB:
      print([child for child in token.head.children]) # Tree navigation
print(subjs, headVerbs, dobjs)

```

#Output the players scoring points at games

```python
scoring_words = ["try","tries","point","points","score","scored","scoring"]
number_words = {"a":"1", "an":"1","one":"1","two":"2","three":"3","four":"4","five":"5","six":"6","seven":"7","eight":"8","nine":"9","ten":"10"}
for sentence in doc.sents:
  if any(word in sentence.text.lower() for word in scoring_words):
    player = [ent.text for ent in sentence.ents if ent.label_ == "PERSON"]
    if not player:
      player = [token.text for token in sentence if token.pos_ == "PROPN"]
    scoring = []
    for i, token in enumerate(sentence):
      if token.text.lower() in scoring_words:
        if i>0:
          previous = sentence[i-1].text.lower()
          if previous in number_words:
            number = number_words[previous]
          elif previous.isdigit():
            number = previous
          else:
            number = "1"
        else:
          number = "1"
        scoring.append(number + " " +token.text)
    if player and scoring:
      print(sentence.text.strip())
      print(player[-1], "←", "".join(scoring))
```

#Output the players receving penalty cards at games

```python
card_words = {"yellow","red","card"}
for sentence in doc.sents:
  if any(word in sentence.text.lower() for word in card_words):
    player = [ent.text for ent in sentence.ents if ent.label_ == "PERSON"]
    if not player:
      player = [token.text for token in sentence if token.pos_ == "PROPN"]
    text = sentence.text.lower()
    if "yellow card" in text:
      card = "Yellow card"
    elif "red card" in text:
      card = "Red card"
    else:
      card = None
    if player and card:
      print(sentence.text.strip())
      print(player[-1], "←", card)
```
